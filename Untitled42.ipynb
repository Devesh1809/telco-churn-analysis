```python
import pandas as pd

# 1. Missing values ani duplicates
nulls = df.isnull().sum()
print("Nulls:\n", nulls[nulls > 0])
print("Duplicate rows:", df.duplicated().sum())

# 2. Text columns madhe nulls bhar
cat_cols = df.select_dtypes(include='object').columns
df[cat_cols] = df[cat_cols].fillna('Not Applicable')

# 3. TotalCharges numeric kar
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce').fillna(0)

# 4. Churn flag (0/1)
df['ChurnFlag'] = (df['ChurnLabel'] == 'Yes').astype(int)
print("Churn rate:", round(df['ChurnFlag'].mean(), 3))

# 5. Tenure group (column cha naav automatic shodh)
tenure_col = [c for c in df.columns if 'tenure' in c.lower()][0]
monthly_col = [c for c in df.columns if 'monthly' in c.lower()][0]
print("Tenure column:", tenure_col, "| Monthly column:", monthly_col)

df['TenureGroup'] = pd.cut(df[tenure_col],
    bins=[-1, 12, 24, 48, 72],
    labels=['0-12', '13-24', '25-48', '49-72'])
print(df['TenureGroup'].value_counts())

# 6. Save ani download (Power BI sathi)
df.to_csv('clean_churn.csv', index=False)
from google.colab import files
files.download('clean_churn.csv')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

def churn_by(col):
    return df.groupby(col, observed=True)['ChurnFlag'].mean().mul(100).round(1).sort_values(ascending=False)

for col in ['Contract', 'TenureGroup', 'InternetType', 'PaymentMethod']:
    print(f"\nChurn % by {col}")
    print(churn_by(col))

# Chart 1: Contract-wise churn
churn_by('Contract').plot(kind='bar', color='tomato')
plt.title('Churn % by Contract')
plt.ylabel('Churn %')
plt.show()

# Chart 2: Tenure group-wise churn
df.groupby('TenureGroup', observed=True)['ChurnFlag'].mean().mul(100).plot(kind='bar', color='steelblue')
plt.title('Churn % by Tenure Group')
plt.ylabel('Churn %')
plt.show()

# Chart 3: Monthly charge vs churn
sns.boxplot(x='ChurnLabel', y='MonthlyCharge', data=df)
plt.title('Monthly Charge: Churned vs Stayed')
plt.show()

# Chart 4: Top 10 churn reasons
df[df['ChurnFlag']==1]['ChurnReason'].value_counts().head(10).plot(kind='barh')
plt.title('Top 10 Churn Reasons')
plt.gca().invert_yaxis()
plt.show()
```
